# echoX - 02: Model Inference & Latency Profiling

Benchmarking AASIST neural forward-pass latency and confidence score distributions.

```python
import sys
import time
import numpy as np
import torch

sys.path.append("..")
from backend.app.services.detector import DETECTOR_SERVICE
from backend.app.core.config import AUDIO_CONFIG

print(f"Active Device: {DETECTOR_SERVICE.device}")
```

```python
# Latency benchmark
dummy_window = np.random.normal(0, 0.2, AUDIO_CONFIG.SAMPLE_WINDOW).astype(np.float32)
latencies = []

for _ in range(30):
    res = DETECTOR_SERVICE.detect(dummy_window)
    latencies.append(res["inference_time_ms"])

latencies = np.array(latencies)
print(f"Benchmark over 30 windows:")
print(f"P50 (Median): {np.percentile(latencies, 50):.2f} ms")
print(f"P95:          {np.percentile(latencies, 95):.2f} ms")
```
