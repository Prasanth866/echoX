# echoX - 03: Risk Fusion Engine & Decision Boundaries

Visualizes the 3-tier security policy boundaries (Low/Medium/High Risk) and frame-by-frame risk assessments.

```python
import sys
import numpy as np
import matplotlib.pyplot as plt

sys.path.append("..")
from backend.app.core.risk_engine import RISK_ENGINE
from backend.app.core.config import RISK_CONFIG

print(f"Risk Engine Thresholds: Low <= {RISK_CONFIG.LOW_MAX}, Med <= {RISK_CONFIG.MED_MAX}, High <= {RISK_CONFIG.HIGH_MAX}")
```

```python
# Frame-by-frame score inspection
probabilities = np.linspace(0.0, 1.0, 100)
risk_scores = [RISK_ENGINE.compute_risk_score(p) for p in probabilities]

plt.figure(figsize=(10, 5))
plt.plot(probabilities, risk_scores, label="Impersonation Risk Score", color="#3b82f6", linewidth=2.5)

# Add decision bands
plt.axhspan(0, 30, color="#10b981", alpha=0.15, label="Low Risk (ALLOW)")
plt.axhspan(30, 60, color="#f59e0b", alpha=0.15, label="Medium Risk (CHALLENGE_MFA)")
plt.axhspan(60, 100, color="#ef4444", alpha=0.15, label="High Risk (BLOCK_ALERT)")

plt.title("echoX: 3-Tier Impersonation Risk Decision Boundaries")
plt.xlabel("Model Spoof Probability")
plt.ylabel("Enterprise Risk Score (0-100)")
plt.ylim(0, 105)
plt.legend(loc="upper left")
plt.grid(True, alpha=0.3)
plt.show()
```
